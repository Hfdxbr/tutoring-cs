```python
N, *data = [[int(x) for x in line.strip().split()] for line in open('26_14595.txt')]
data.sort(key = lambda x: x[1])
time = 0
counter = 0
tech = 0
time_tech = 0
nado = []
for i in range(1, len(data)):
    if data[i - 1][1] == data[i][0] and data[i - 1][0] > time:
        counter += 2
        tech += 2
        time += max(data[i - 1][1], data[i][1])
        nado.extend(data[i - 1:i + 1])
    if data[i - 1][1] > data[i][0] and data[i - 1][0] > time:
        counter += 1
        tech += 1
        time += data[i - 1][1]
        nado.append(data[i - 1])
print(counter, nado, time, time_tech) # 26 36
```
