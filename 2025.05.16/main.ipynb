```python
from collections import defaultdict

# N, *data = [[int(x) for x in line.strip().split()] for line in open('26_9847.txt')]
counter = defaultdict(int)
nado = {}
s = 0

for tin, tout in data:
    counter[tin] += 1
    counter[tout] -= 1

for key in sorted(counter):
    delta = counter[key]
    if delta != 0:
        s += delta
        nado[key] = s

max_amount = max(nado.values())
count = sum(v == max_amount for v in nado.values())
print(count, max_amount)
```

```python
def their(a):
    a.sort()
    kmx = 0
    mx = 0
    last = 0
    shop = []
    for i in a:
        j = 0
        while j < len(shop):
            if shop[j] < i[0]:
                del shop[j]
            else:
                j += 1
        shop.append(i[1])
        if len(shop) > mx:
            mx = len(shop)
            kmx = 1
        elif len(shop) == mx and last != mx:
            kmx += 1
        last = len(shop)
    return kmx, mx
```
