```python
def F(x, y, m):
    if x + y <= 69:
        return m % 2 == 0
    if m == 0:
        return 0
    h = [F(x - 5, y + 2, m - 1), F(x + 2, y - 5, m - 1), F(x, y // 2, m - 1), F(x // 2, y, m - 1)]
    return any(h) if m % 2 != 0 else all(h)
print(max([x for x in range(51, 500) if F(35, x, 2)])) # 70

```

```python
def F(x, y):
    if x > y:
        return 0
    if x == y:
        return 1
    return F(x + 2, y) + F(x + 3, y) + F(x + 5, y)
print(F(5, 11) * F(11, 22) * F(22, 33) * F(33, 37))
# 512
```
