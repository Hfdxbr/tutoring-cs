```python
from itertools import *

def F(x,y,z,w):
    return (not(w <= z)) or (x == y) or x

for a1, a2, a3, a4, a5, a6, a7 in product([0, 1], repeat = 7):
    t = [(1, a1, a2, a3), (0, 1, 0, a4), (a5, 0, a6, a7)]
    if len(t) == len(set(t)):
        for p in permutations('xyzw'):
            if [F(**dict(zip(p, r))) for r in t] == [0, 0, 0]:
                print(*p)
# wzxy
```

```python
def F(x):
    s = ''
    while x > 0:
        s = str(x % 4) + s
        x //= 4
    return s

minn = 99999999
nado = 0
for x in range(1, 2000):
    s = F(4 ** 210 + 4 ** 110 - x).count('0')
    if s <= minn:
        minn = s
        nado = x
print(nado)
```

```python
def F(A1, A2, x):
    return (not(A1 <= x <= A2)) <= ((66 <= x <= 75) == (71 <= x <= 85))

minn = 99999999999
for A1 in range(500):
    for A2 in range(A1 + 1, 500):
        if all(F(A1, A2, x) for x in range(500)):
            minn = min(minn, A2 - A1)
print(minn)
```

```python
data = [int(x) for x in open('17.txt')]
maxx = max(x for x in data if len(str(abs(x))) == 4 and abs(x) % 10 == 5)
maxx_2 = -9999999999
counter = 0
for i in range(1, len(data)):
    pair = data[i - 1:i + 1]
    if sum(len(str(abs(x))) == 4 and abs(x) % 10 == 5 for x in pair) >= 1 and sum(pair) > maxx:
        counter += 1
        maxx_2 = max(maxx_2, sum(pair))
print(counter, maxx_2)
```
