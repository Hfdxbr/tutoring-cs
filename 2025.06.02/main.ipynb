```python
storage = [0 for _ in range(K)]

def get_storage(time):
    for i, s in enumerate(storage):
        if s <= time:
            return i
counter = 0
latest_time = 0
for time, duration in sorted(data):
    if (i:=get_storage(time)) is None:
        counter += 1
    else:
        storage[i] = time + duration
        latest_time = storage[i]

print(counter, latest_time)
```

```python
# Read input
K, T = map(int, input().split())
N = int(input())
buses = [tuple(map(int, input().split())) for _ in range(N)]

last_visit = [-1] * K  # Initialize with -1 (no visits)
total_buses = 0
max_last_stop = 0

for start, end in buses:
    current_stop = 1
    current_time = start
    visited = False
    while current_time <= end and current_stop <= K:
        if last_visit[current_stop - 1] < current_time:
            # Bus stops here
            visited = True
            last_visit[current_stop - 1] = current_time
            max_last_stop = max(max_last_stop, current_stop)
            current_time += T
            current_stop += 1
        else:
            # Skip this stop
            current_time += T
            current_stop += 1
    if visited:
        total_buses += 1

print(total_buses, max_last_stop)
```

```python
(N,M,K), *data = [list(map(int,x.split())) for x in open('13_26.txt').readlines()]
data.sort(reverse=True)
cur_row = data[0][0]
cur_places = set()
above_places = set()
def check(places):
    if len(places) <= 1:
        return
    for i in range(1, len(places)):
        if places[i - 1] == places[i] + 1:
            return places[i]
            

for row, place in data:
    if row == cur_row:
        cur_places.add(place)
    else:
        possible = check(sorted(above_places - cur_places, reverse=True))
        if possible is not None:
            print(cur_row, possible)
            break
        above_places = cur_places
        cur_places = {place}
        cur_row = row
```
