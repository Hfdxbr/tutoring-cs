```python
# 2 +
from itertools import *
def F(x, y, z, w):
    return (not(not(x) or y)) or (y == z) or (not(w))

for a1, a2, a3, a4 in product([0, 1], repeat = 4):
    t = [(a1, a2, 0, 0), (1, 0, a3, 0), (1, 0, 1 , a4)]
    if len(t) == len(set(t)):
        for p in permutations('xyzw'):
            if [F(**dict(zip(p, r))) for r in t] == [0, 0, 0]:
                print(*p)
# wzyx
```

```python
# 16 +
from math import prod
data = {}
def F(n):
    if n not in data:
        if n > 2000:
            data[n] = 16
        else:
            data[n] = 2 * F(n + 3)
    return data[n]
print(prod(int(x) for x in str(F(50) // F(110)).replace('0', '1')))
# 6720
```

```python
# 17 +
data = [int(x) for x in open('17.txt')]
summ = sum(x for x in data if len(str(abs(x))) == 4)
minn = 99999999
counter = 0
for i in range(2, len(data)):
    pair = data[i - 2:i + 1]
    if sum(len(str(abs(x))) == 3 for x in pair) == 2 and prod(pair) > summ:
        counter += 1
        minn = min(minn, prod(pair))
print(counter, abs(minn))
# 566 1462000
```

```python
# 19 +
def F(x, m):
    if x >= 333:
        return m % 2 == 0
    if m == 0:
        return 0
    h = [F(x + 3, m - 1), F(x + 8, m - 1), F(x * 2, m - 1)]
    return any(h) if m % 2 != 0 else all(h)
print([x for x in range(1, 333) if F(x, 2)]) 
# 164
```

```python
# 23 +
from math import ceil
def F(x, y):
    if x < y or x == 26 or x == 30:
        return 0
    if x == y:
        return 1
    return F(x - 3, y) + F(ceil(x / 2), y)
print(F(69, 14))
# 5
```

```python
# 24
s = open('24.txt').read().replace('-', '*')
maxx = 0
counter = 0
for x in s.split('**'):
    if any(len(n) > 1 and n[0] == '0' for n in x.split('*')):
        continue
    n = len(x.split('*'))
    counter += (n * (n - 1)) // 2
print(counter)
# 987*654 => [987*654, 987*65, 987*6, 87*654, 87*65, 87*6, 7*654, 7*65, 7*4]
```

```python
import re

counter = 0
pattern = re.compile(r"((0|[1-9][0-9]*)([-\*](0|[1-9][0-9]*))+)")
s = open("24.txt").read()
groups = re.findall(pattern, s)
for match in groups or []:
    s = match[0].replace("-", "*")
    digits = len(s) - s.count("*")
    for group in s.split("*"):
        digits -= len(group)
        skiped_zeros = group.count("0") - group.endswith("0")
        counter += (len(group) - skiped_zeros) * digits

print(counter)
# 980*402*77 // 8 digits
# [980]*402*77  // 3*5 seqs
# [402]*77 // 2*2 seqs
```
