```python
from collections import defaultdict

s = open("24_22446.txt").readline()
data = [defaultdict(list)]
for i, c in enumerate(s):
    if s[i : i + 3] == "LND":
        data.append(defaultdict(list))
    data[-1][c].append(i)
maxim = 0
for i in range(len(data) - 10000):
    for c in set(data[i]) & set(data[i + 10000]):
        maxim = max(max(data[i + 10000][c]) - min(data[i][c]), maxim)
print(maxim + 1)
```

```python
# 830019
s = open('24_22446.txt').read().split('LND')
maxx = 0
for i in range(len(s) - 10_000):
    x = 'LND'.join(s[i:i + 10_001])
    cur_maxx = len(x)
    if i == 0:
        if x[0] == 'N':
            cur_maxx += 2 
        if x[0] == 'L':
            cur_maxx += 1 
    elif i + 10_001 == len(s):
        if x[-1] == 'N':
            cur_maxx += 2
        if x[-1] == 'D':
            cur_maxx += 1 
    else:
        cur_maxx += 4
    if cur_maxx > maxx:
        maxx = cur_maxx

print(maxx)
```
