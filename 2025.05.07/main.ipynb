```python
from itertools import *
def F(x, y, z, w):
    return (not(x) or not(w) or not(z) or y) and (not(x) or w or z or y) and (x or w or not(z) or not(y))

for a1, a2, a3, a4 in product([0, 1], repeat = 4):
    t = [(0, 1, 1, 0), (a1, 1, 0, 1), (0, a2, a3, a4)]
    if len(t) == len(set(t)):
        for p in permutations('xyzw'):
            if [F(**dict(zip(p, r))) for r in t] == [0, 0, 0]:
                print(*p)
# wzyx
```

```python
data = [int(x) for x in open('17.txt')]
counter = 0
maxx = -99999999999
pairs_r = []
pairs_i = []
def F(a1):
    return (len(str(abs(a1))) == 3), (a1 % 75 == 0)
for i in range(1, len(data)):
    a1, a2 = data[i - 1:i + 1]
    if [sum(p) for p in zip(F(a1), F(a2))] == [1, 1]:
        counter += 1
        maxx = max(maxx, abs(abs(a1) - abs(a2)))
        pairs_r.append([a1,a2])
print(counter, maxx)

counter = 0
maxx = -99999999999
for i in range(1, len(data)):
    pair = data[i - 1:i + 1]
    a = [x for x in pair if len(str(abs(x))) == 3]
    b = [x for x in pair if x % 75 == 0]
    if len(a) == 1 and len(b) == 1 and a != b:
        counter += 1
        maxx = max(maxx, abs(abs(pair[0]) - abs(pair[-1])))
        pairs_i.append(pair)
print(counter, maxx)
# 39 9559
```

```python
def F(x, m):
    if x >= 111:
        return m % 2 == 0
    if m == 0:
        return 0
    h = [F(x + 1, m - 1), F(x + 4, m - 1), F(x * 2, m - 1)]
    return any(h) if m % 2 != 0 else all(h)

print([x for x in range(1, 111) if F(x * 2, 1)])
# 28
```

```python
def F(x, y, p):
    if x < y:
        return 0
    if x == y and p == 'C':
        return 1
    return F(x - 6, y, '') + F(x - 1, y, '') + F(x // 2, y, 'C')
print(F(57, 15, ''))
# 736
```

```python
def F(x):
    M = []
    for y in range(2, int(x ** 0.5) + 1):
        if x % y == 0:
            M.append(y)
            M.append(x // y)
    M = set(M)
    S = sum([z for z in M if all(z % g != 0 for g in range(2, int(z ** 0.5) + 1))])
    return S if S > 50_000 and S % 1000 == 123 else None
for x in range(1_200_000, 1_250_000):
    if F(x):
        print(x, F(x))
# 1208968 151123
# 1212242 606123
# 1224484 306123
# 1228484 307123
# 1236484 309123
# 1241373 59123
```
